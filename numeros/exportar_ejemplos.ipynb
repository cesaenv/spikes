# Exportar ejemplos de prueba

Guarda **un dígito del test de cada clase** en `modelo/ejemplos.json`. La página de dígitos lo usa para el botón *Dígito de prueba*: en lugar de dibujar, muestra una imagen real de MNIST y la red intenta reconocerla.

Hay que ejecutarlo desde la carpeta `numeros/`. Solo hace falta una vez, salvo que cambies el dataset. No depende del entrenamiento, así que se puede ejecutar antes o después de `entrenar.ipynb`.

```
import json, os
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"
import numpy as np, keras
```

Para cada dígito de la lista (en un orden fijo, para que el botón siempre salga igual) se toma la primera imagen del test con esa etiqueta. Los píxeles se guardan como enteros de 0 a 255 y aplanados, sin normalizar: la normalización se hace en el navegador.

```
(_, _), (x, y) = keras.datasets.mnist.load_data()
ej = []
for d in [7, 2, 1, 0, 4, 9, 5, 3, 8, 6]:
    i = int(np.where(y == d)[0][0]); ej.append({"digito": d, "px": x[i].flatten().tolist()})
json.dump(ej, open("modelo/ejemplos.json", "w"), separators=(",", ":"))
print(len(ej), "ejemplos guardados")
```
